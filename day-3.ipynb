```python
import pandas as pd
```

```python
df = pd.read_excel('/content/CineMetrics_Dataset.xlsx')
```

```python
df["Duration_Min"] = df["Duration_Min"].fillna(df["Duration_Min"].mean())
df.info()
```

```python
df["Budget_Million_USD"] = df["Budget_Million_USD"].fillna(df["Budget_Million_USD"].mean())
df.info()
```

```python
df["Revenue_Million_USD"] = df["Revenue_Million_USD"].fillna(df["Revenue_Million_USD"].mean())
df.info()
```

```python
df["IMDB_Rating"] = df["IMDB_Rating"].fillna(df["IMDB_Rating"].mean())
df.info()
```

```python
df["Votes_Thousands"] = df["Votes_Thousands"].fillna(df["Votes_Thousands"].mean())
df.info()
```

```python
df["Critic_Score"] = df["Critic_Score"].fillna(df["Critic_Score"].mean())
df.info()
```

```python
df["Director_Experience_Years"] = df["Director_Experience_Years"].fillna(df["Director_Experience_Years"].mean())
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
numeric_cols = ['Duration_Min', 'Budget_Million_USD', 'Revenue_Million_USD',
                'IMDB_Rating', 'Votes_Thousands', 'Critic_Score',
                'Director_Experience_Years']
```

```python
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())
```

```python
df[numeric_cols].isnull().sum()
```

```python
df.isnull().sum()
```

```python
before_rows = df.shape[0]
before_rows
```

```python
df = df.dropna()
```

```python
after_rows = df.shape[0]
print('Rows before:', before_rows)
print('Rows after: ', after_rows)
print('Dropped:    ', before_rows - after_rows)
```
